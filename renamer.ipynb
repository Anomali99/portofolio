```python
from typing import Optional, List
import os

urls = []
```

```python
def renamefile(path: str, source: str) -> None:
    file_list = os.listdir(path)
    for file in file_list:
        file_path = os.path.join(path, file)
        new_name = file.replace(" ", "-").replace("_", "-").replace("---", "-").replace("--", "-").replace("'", "").lower()
        new_path = os.path.join(path, new_name)
        os.rename(file_path, new_path)
        if os.path.isdir(new_path):
            renamefile(new_path, source + "/" + file)
        else:
            print(new_path)
            urls.append(source + "/" + file)
```

```python
renamefile(
    path="/home/anomali99/Documents/Portofolio/public/docs/artikel",
    source="https://anomali99.my.id"
)
```

```python
for url in urls:
    print(url)
```
